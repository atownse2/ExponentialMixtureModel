```python
import ROOT
import emm

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Get data
data_tree = emm.get_data(sort_and_index=True, tree=True)

x = ROOT.RooRealVar("x", "Diphoton Mass [GeV]", 500, 4000)
index=ROOT.RooRealVar("index", "index", 0, 0, 1e6)

data = ROOT.RooDataSet("mgg", "mgg", ROOT.RooArgSet(x, index), ROOT.RooFit.Import(data_tree))
```

```python
pset = {
    'raw_rate_0': (0.3, 2.5, 20),
    'raw_rate_1': (0.3, 2.5, 20),
}

def model_primitive(x, data):
    model = emm.ExponentialMixtureModel(x, 2, data_mean=data.mean(x),)
    return model

df = emm.profile_model(x, model_primitive, data, pset)
fig, axes = emm.plot_pair_profiles(df, pset)
```

```python

pset = {
    'raw_rate_0': (0.3, 2.5, 20),
    'raw_rate_1': (0.3, 2.5, 20),
    'raw_rate_2': (0.3, 2.5, 20),
}

def model_primitive(x, data):
    model = emm.ExponentialMixtureModel(x, 3, data_mean=data.mean(x),)
    return model

df = emm.profile_model(x, model_primitive, data, pset)
fig, axes = emm.plot_pair_profiles(df, pset)
```

```python
def model_primitive(x, data):
    model = emm.ExponentialMixtureModel(x, 3, data_mean=data.mean(x),)
    v0 = model.get_param("raw_weight_0")
    v0.setVal(0.001)
    v0.setRange(0,0.1)

    return model

df = emm.profile_model(x, model_primitive, data, pset)
fig, axes = emm.plot_pair_profiles(df, pset)
```

```python
def model_primitive(x, data):
    model = emm.ExponentialMixtureModel(x, 3, data_mean=data.mean(x),)
    v0 = model.get_param("raw_weight_0")
    v0.setVal(0.001)
    v0.setRange(0,0.01)

    return model

df = emm.profile_model(x, model_primitive, data, pset)
fig, axes = emm.plot_pair_profiles(df, pset)
```

```python

pset = {
    'raw_rate_0': (0.3, 2.5, 30),
    'raw_rate_1': (0.3, 2.5, 30),
    'raw_rate_2': (0.3, 2.5, 30),
}

def model_primitive(x, data):
    model = emm.ExponentialMixtureModel(x, 3, data_mean=data.mean(x), ordered_weights=True)
    return model

df = emm.profile_model(x, model_primitive, data, pset)
fig, axes = emm.plot_pair_profiles(df, pset)
```

```python
def model_primitive(x, data):
    model = emm.ExponentialMixtureModel(x, 3, data_mean=data.mean(x), ordered_weights=True)
    return model

model = model_primitive(x, data)
emm.fit_and_plot(model, data, x, print_pars=False, minos=True)
for i in range(model.n_components):
    print(f"Component {i}:")
    emm.print_par(model.weights[i])
    emm.print_par(model.raw_rates[i])
```
